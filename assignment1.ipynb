### Overview
Single-view metrology corresponds to a set of techniques for making 3-D measurements from a single 2-D image. Such measurements have several applications including, for example, measuring the height of a suspect in a crime-scene image. We will consider a simple scenario of single-view metrology estimating 3-D measurements of objects in a rendered 2-D image.

### Description

Consider the image shown in Figure 1. You are given the following information about the 3-D scene. There are three poles colored yellow, green, and blue, and a ground plane in this scene. The three poles are perpendicular to the ground plane. The ground plane is parallel to the X-Z plane, and the camera's optical axis is parallel to the Z-axis. Standing at 4 meters, the yellow and green poles are of the same height in the 3-D scene. The distance between these two poles is 17 meters. The green and blue poles are at the same, but unknown, distance from the camera. Finally, the resolution of the image sensor of the camera is 53,333 pixels per meter.

Recall the perspective projection equation that tells us the relationship between a point's location in the 3-D world (X,Y,Z), the camera focal length (f), and the projection of that point in the image (x,y):  
$$\frac{x}{X}=\frac{f}{Z}$$
and
$$\frac{y}{Y}=\frac{f}{Z}$$

Now we are able to find the distances along the optical axis to the green and yellow poles. Let:
- $Z^{(y)}$ denote the real-world distance (in meters) from the camera to the yellow pole along the optical axis.
- $h$ denote the height of the camera above the ground (in meters).
- $y_\text{bot}^{(y)}$ denote the $y$-position of the bottom of the yellow pole in the image plane (in meters).
- $y_\text{bot}^{(g)}$ denote the $y$-position of the bottom of the green pole in the image plane (in meters).
<br>

Since the bottom of each pole is exactly on the ground, these endpoints both satisfy $Y = -h$. The focal length is constant, so we can set the perspective projection equations for the green and yellow poles equal to each other:

$$Z^{(y)}\left(\frac{y_\text{bot}^{(y)}}{- h}\right) = f = \left( Z^{(y)} + 17 \right)\left(\frac{y_\text{bot}^{(g)}}{- h}\right).$$

The height $h$ cancels out, so distributing gives
$$Z^{(y)}y_\text{bot}^{(y)} = Z^{(y)}y_\text{bot}^{(g)} + 17y_\text{bot}^{(g)}.$$
Finally, we can combine like terms to find
$$Z^{(y)} = \frac{ 17y_\text{bot}^{(g)} }{  y_\text{bot}^{(y)} -  y_\text{bot}^{(g)} }.$$


```python
y_bot_green  = scaled_pole_endpoints['green' ][0, 1]
y_bot_yellow = scaled_pole_endpoints['yellow'][0, 1]

yellow_Z = (17 * y_bot_green) / (y_bot_yellow - y_bot_green)
green_Z  = yellow_Z + 17

print(f'Distance to yellow pole: {yellow_Z:.5f} m.')
print(f'Distance to green  pole: { green_Z:.5f} m.')
```

Next, we need to find of the camera above the ground plane. For this calculation, we can use two endpoints from the same pole, which will share the same values for $X$ and $Z$. Since $f$ and $Z$ are both fixed here, we can again set the perspective projection ratios equal to each other:

$$\frac{f}{Z_\text{bot}} = \frac{f}{Z_\text{top}} \implies \frac{ y_\text{bot} }{ Y_\text{bot} -h } = \frac{ y_\text{top} }{ Y_\text{top} -h}.$$


Again, we know that the bottom endpoints must be on the ground, and each of the yellow and green poles are 4m tall. Plugging these values in gives
$$\frac{ y_\text{bot} }{ -h } = \frac{ y_\text{top} }{ 4 -h} \implies h = \frac{4y_\text{bot}}{y_\text{bot} - y_\text{top}}.$$

I defined a function to programmatically compute the height using a specific pole:

```python
def find_height_from_endpoints(endpoints_dict: dict, pole_color: str) -> float:
    assert pole_color in ('green', 'yellow')

    y_bottom = endpoints_dict[pole_color][0,1]
    y_top    = endpoints_dict[pole_color][1,1]

    return float((4 * y_bottom) / (y_bottom - y_top))
```

#### Computing the Focal Length

Now that we know the $X, Y$ and $Z$ coordinates for the yellow and green poles, we can use the perspective projection equations to find $f$:
$$f = \frac{xZ}{X} = \frac{yZ}{Y}.$$

```python
def compute_focal_length(x: float, X: float, Z: float) -> float:
    return float(Z * x / X)
```

*Extra notes for those that are interested or curious on practical applications beyond this course*:

Coordinate systems:

1. ECEF - Earth-Centered, Earth-Fixed
  - A global 3D Cartesian coordinate system fixed to the Earth.
  - Origin is at the center of the Earth.
  - Axes are:
    - X → intersection of Equator & Prime Meridian (0° lat, 0° lon)
    - Y → Equator & 90°E longitude
    - Z → North Pole

    Units: Meters

    Why it exists:
    - GPS (lat, lon, alt) is defined from ground level, but ground level varies throughout the earth, and inconvenient for geometry.
    - ECEF lets you do straight-line vector math in 3D.

2. ENU — East, North, Up (Local Tangent Plane)
  - A local coordinate system anchored at the camera location.
  - Think: “flat ground under the camera”.
  - Axes are:
    - X → East
    - Y → North
    - Z → Up
    - Origin is the camera's GPS position
  - Why it exists
    - ECEF values are huge (~6,371 km).
    - ENU gives small, meaningful relative coordinates:
      - “10 m east, 5 m north, 2 m up from the camera”
    - ECEF → ENU is a rotation + translation.

3. CAM — Camera Coordinate Frame (OpenCV)
    - The coordinate system used by pinhole camera models.
    - Origin is at the camera optical center.
    - Axis (OpenCV convention)
      - X → right in the image
      - Y → down in the image
      - Z → forward (out of the camera)


4. How they connect:
  GPS (lat, lon, alt)
  ->
  ECEF (global XYZ, meters)
  ->
  ENU (local XYZ, meters, camera-centered)
  ->
  CAM (camera XYZ, meters)
  ->
  IMAGE (pixels)

## Get Calibration with CV2

```python
from pyproj import Transformer

def estimate_camera_params(camera_gps, poi_gps, poi_xy, frame_size, calibration_flags, intrinsics_estimate, distortion_estimate):
    # Convert camera GPS coordinates to ECEF
    transformer_geodetic_to_ecef = Transformer.from_crs("epsg:4979", "epsg:4978", always_xy=True)
    # Note: pyproj expects lon,lat,alt order
    eX, eY, eZ = transformer_geodetic_to_ecef.transform(camera_gps[1], camera_gps[0], camera_gps[2])
    cam_ecef = np.array([eX, eY, eZ]).T#/1000  # Shape: (3,)
    eX, eY, eZ = transformer_geodetic_to_ecef.transform(poi_gps[:,1], poi_gps[:,0], poi_gps[:,2])
    poi_ecef = np.vstack((eX, eY, eZ)).T#/1000  # Shape: (3, N)
    poi_enu = ecef_to_enu(cam_ecef, poi_ecef, camera_gps[0],camera_gps[1])  # Shape: (N,3)

    # Put everything in contiguous arrays with (N, 3), not (3, N)
    object_points = np.ascontiguousarray(poi_enu.astype(np.float32))  # (N, 3)
    image_points = np.ascontiguousarray(poi_xy.astype(np.float32))      # (N, 2)

    criteria = (
        cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER,
        100000,    # Maximum 10,000 iterations (vs default ~30)
        1e-12     # Extremely tight convergence (vs default 1e-6)
        )


    #Let OpenCV estimate all distortion parameters
    ret, camera_matrix, dist_coeffs, rvecs, tvecs = cv2.calibrateCamera(
        [object_points], [image_points], frame_size,
        intrinsics_estimate,
        distortion_estimate,
        flags=calibration_flags,
        criteria=criteria
    )

    #convert to canonical representation
    T = tvecs[0]
    R, _ = cv2.Rodrigues(rvecs[0])

    #return  intrinsics_estimate, distortion_estimate, R, T, cam_ecef
    return  camera_matrix, dist_coeffs, R, T, cam_ecef
```

```python
def ecef_to_enu(origin_ecef, points_ecef, cam_lat, cam_lon):

    lat0 = np.radians(cam_lat)
    lon0 = np.radians(cam_lon)

    # Rotation matrix from ECEF to ENU
    R = np.array([
        [-np.sin(lon0),              np.cos(lon0),             0],
        [-np.sin(lat0)*np.cos(lon0), -np.sin(lat0)*np.sin(lon0), np.cos(lat0)],
        [ np.cos(lat0)*np.cos(lon0),  np.cos(lat0)*np.sin(lon0), np.sin(lat0)]
    ])

    # Subtract camera position (origin)
    delta = points_ecef - origin_ecef[None,:]
    return (R @ delta.T).T  # Shape (N, 3)

def gps_to_camxy(lats, lons, alts, cam_ecef, cam_k, cam_r, cam_t, camera_gps, distortion):
    """
    Simplified GPS to Camera conversion.
    Returns X, Y pixels and Z depth for ALL points without masking.
    """
    # 1. Convert GPS to ECEF
    transformer_geodetic_to_ecef = Transformer.from_crs("epsg:4979", "epsg:4978", always_xy=True)
    eX, eY, eZ = transformer_geodetic_to_ecef.transform(lons, lats, alts)
    ecef_points = np.column_stack((eX, eY, eZ))

    # 2. Convert ECEF to ENU relative to camera
    enu_points = ecef_to_enu(cam_ecef, ecef_points, camera_gps[0], camera_gps[1])

    # 3. Transform to camera coordinate system (ENU to Camxy)
    # This gives us the raw 3D position relative to the lens
    # NOTE: Here we define x-y with (0,0) in the bototm left corner of the image.
    # OPENCV and some other conventions define (0,0) as the TOP left corner of the image.
    points_cam = (cam_r @ enu_points.T + cam_t).T
    cam_depths = points_cam[:, 2] # This is our Z (depth)

    # 4. Project ALL points using OpenCV
    rvec, _ = cv2.Rodrigues(cam_r)
    tvec = cam_t.astype(np.float32)

    # projectPoints handles the perspective division (X/Z, Y/Z)
    image_points, _ = cv2.projectPoints(
        enu_points.astype(np.float32),
        rvec, tvec, cam_k, distortion
    )

    image_points = image_points.reshape(-1, 2)
    image_x = image_points[:, 0]
    image_y = image_points[:, 1]

    return image_x, image_y, cam_depths
```

```python
from shapely.geometry import Polygon as ShapelyPolygon
def calculate_camera_angles(cam_r):
    """
    Calculate camera orientation angles from rotation matrix.

    Returns:
        azimuth: Camera azimuth in degrees (0=North, 90=East, 180=South, 270=West)
        elevation: Camera elevation/pitch in degrees (positive=up, negative=down)
        roll: Camera roll in degrees (rotation around forward axis)
    """
    # Camera directions
    #right = cam_r[:, 0]    # Camera right direction in ENU (for Roll)
    #down = cam_r[:, 1]     # Camera down direction in ENU (for Roll)
        # The rotation matrix transforms from ENU to camera coordinates
    # To get camera orientation in ENU, we need the inverse (transpose for orthogonal matrix)
    R_camera_to_enu = cam_r.T

    # Camera forward direction in ENU coordinates
    # In camera space, forward is [0, 0, 1], so in ENU it's the 3rd column of R^T
    forward_enu = R_camera_to_enu[:, 2]

    # Azimuth: angle in horizontal plane (from North)
    # In ENU: North is +Y, East is +X
    azimuth = np.degrees(np.arctan2(forward_enu[0], forward_enu[1]))
    if azimuth < 0:
        azimuth += 360

    # Elevation: angle from horizontal plane
    horizontal_distance = np.sqrt(forward_enu[0]**2 + forward_enu[1]**2)
    elevation = np.degrees(np.arctan2(forward_enu[2], horizontal_distance))

    return (azimuth.item(), elevation.item())

def create_visibility_cone_polygon(cam_lat, cam_lon, cam_r, max_distance_km, fov_h_deg, fov_v_deg=None):
    """
    Create a polygon representing the camera's visibility cone on the ground.

    Args:
        cam_lat, cam_lon: Camera location
        cam_r: Camera rotation matrix
        max_distance_km: Maximum visibility distance
        fov_h_deg: Horizontal field of view in degrees
        fov_v_deg: Vertical FOV (optional, used to limit distance based on depression angle)

    Returns:
        ShapelyPolygon: Polygon representing visibility cone
    """
    # Get camera azimuth
    azimuth, elevation = calculate_camera_angles(cam_r)

    # Create arc points
    num_points = 50
    half_fov = fov_h_deg / 2

    # Angles for the arc (from -half_fov to +half_fov relative to camera azimuth)
    angles = np.linspace(azimuth - half_fov, azimuth + half_fov, num_points)

    # Convert to radians
    angles_rad = np.radians(angles)

    # Calculate arc points at max distance
    arc_points = []
    for angle in angles_rad:
        # Calculate point at max_distance from camera
        lat_offset = np.cos(angle) * max_distance_km / 111.0  # ~111 km per degree latitude
        lon_offset = np.sin(angle) * max_distance_km / (111.0 * np.cos(np.radians(cam_lat)))

        point_lat = cam_lat + lat_offset
        point_lon = cam_lon + lon_offset
        arc_points.append((point_lon, point_lat))  # Shapely uses (lon, lat) order

    # Add camera position to close the polygon
    arc_points.append((cam_lon, cam_lat))

    return ShapelyPolygon(arc_points)


def add_visibility_cone_to_map(ax, cam_lat, cam_lon, cam_r, max_distance_km,
                              fov_h_deg, fov_v_deg=None, color='yellow', alpha=0.3,
                              edge_color='orange', linewidth=2, arrow=True, arrow_label=True):
    """
    Add visibility cone to an existing map plot.

    Args:
        ax: Matplotlib axes (in Web Mercator projection)
        Other args same as create_visibility_cone_polygon
    """
    # Create the cone polygon
    cone_polygon = create_visibility_cone_polygon(cam_lat, cam_lon, cam_r, max_distance_km, fov_h_deg, fov_v_deg)

    # Convert to GeoDataFrame
    gdf_cone = gpd.GeoDataFrame([1], geometry=[cone_polygon], crs='EPSG:4326')

    # Project to Web Mercator for display
    gdf_cone = gdf_cone.to_crs('EPSG:3857')

    # Plot the cone
    gdf_cone.plot(ax=ax, color=color, alpha=alpha, edgecolor=edge_color, linewidth=linewidth)

    #set the axis limits to the cone bounds *must do this before adding arrows*
    cone_bounds = gdf_cone.total_bounds  # returns [minx, miny, maxx, maxy]
    # Add some padding (10% of the extent)
    x_padding = (cone_bounds[2] - cone_bounds[0]) * 0.1
    y_padding = (cone_bounds[3] - cone_bounds[1]) * 0.1
    # Set axis limits to encompass the cone with padding
    ax.set_xlim(cone_bounds[0] - x_padding, cone_bounds[2] + x_padding)
    ax.set_ylim(cone_bounds[1] - y_padding, cone_bounds[3] + y_padding)

    # Add camera direction arrow
    azimuth, _ = calculate_camera_angles(cam_r)
    arrow_length = max_distance_km * 0.9 * 1000  # 30% of max distance in meters

    # Get camera position in Web Mercator
    transformer = Transformer.from_crs("EPSG:4326", "EPSG:3857", always_xy=True)
    cam_x, cam_y = transformer.transform(cam_lon, cam_lat)

    if(arrow):
        # Calculate arrow end point
        arrow_dx = np.sin(np.radians(azimuth)) * arrow_length
        arrow_dy = np.cos(np.radians(azimuth)) * arrow_length

        # Draw arrow
        ax.arrow(cam_x, cam_y, arrow_dx, arrow_dy,
                head_width=arrow_length*0.05, head_length=arrow_length*0.05,
                fc='red', ec='darkred', linewidth=2, zorder=10)

    if(arrow_label):
        # Add text label
        ax.text(cam_x + arrow_dx*1.1, cam_y + arrow_dy*1.1,
                f'Azimuth:{azimuth:.0f}°\n FOV:{fov_h_deg:.0f}°(h),{fov_v_deg:.0f}°(v)', fontsize=8, color='black',
                ha='center', va='bottom', weight='bold',
                bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.5))


    return(ax)
```
